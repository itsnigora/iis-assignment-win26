3. **E3 · The Commitment Decision & Asymmetric Failure Costs**

    - During the retrieval step, the robot faces a choice between a **5 mm test lift near the shelf** or a **high carry directly to the counter**. State the predicted effect, the cost of failure, and the single piece of evidence that justifies choosing the low test lift before committing to the carry.

      **Answer:** The predicted effect of the 5 mm test lift is to check whether the carton can be securely held before transporting it. If the test fails, the carton remains close to the shelf, reducing the risk of damage and spilling. A direct high carry has a higher failure cost because the carton could fall from a greater height. The decisive evidence is whether the carton remains stable during the test lift. Since downward slipping is observed, the robot must adapt its grasp before carrying the carton.
      

4. **E4 · Causal Prediction across Material Variants**

    - Compare pouring milk into coffee versus pancake batter into a pan. Formulate the stopping rule for each liquid using `(:stop-when ...)`, and explain how material dynamics alter when the robot must command the wrist to tilt back.

      **Answer:** For milk: `(:stop-when coffee-milk-level-reaches-target)`
      For pancake batter: `(:stop-when predicted-batter-amount-reaches-target)`
      When pouring milk into coffee, the robot should tilt back as the desired milk quantity is reached because milk responds relatively quickly to changes in container angle. For pancake batter, the robot must start tilting back earlier because its greater viscosity and residual flow can cause additional batter to enter the pan after the stopping motion begins. The central idea is anticipating flow and avoiding overpouring.

5. **E5 · System Decomposition & The Reasoning Loop**

    - Trace the carton slip event through the seven-stage reasoning loop: **Infer → Ground → Predict → Decide → Regulate → Detect & Revise → Verify**. State precisely which step was triggered when downward motion was observed during the 5 mm lift, and how the action designator was developed in response.
  
      **Answer:** **1.Infer:** Interpret Eva's request as serving her usual coffee with milk.
      **2.Ground:** Identify the coffee mug and the opened whole-milk carton (Carton A).
       **3.Predict:** Estimate whether the planned grasp can securely lift the carton.
       **4.Decide:** Choose a low 5 mm test lift instead of a high carry.
       **5.Regulate:** Execute the test lift using controlled gripper force and arm motion.
       **6.Detect & Revise:** Detect the downward slipping motion and revise the action designator to use a more secure grasp, for example by changing the grasp position or permitted grip force.
        **7.Verify:** Repeat the stability check and proceed with pouring and delivery only after the adapted grasp is verified.

6. **E6 · Conceptual Ablation & Evidence of Competence**

    - Perform an ablation experiment. Suppose we remove the **"Check before commitment"** capability and replace it with a fixed default high-side-pinch lift. Predict a specific kitchen environment variant where this ablated system fails, and explain why single-run execution success differs from evidence of general competence across task variants.

      **Answer:** If the **"Check before commitment"** capability is removed, the robot may fail when handling a wet or slippery milk carton. With a fixed high-side-pinch grasp, the carton could slip and fall during a high carry, spilling milk or damaging the environment. A single successful execution only demonstrates that the robot worked under one particular condition. General competence requires the robot to adapt and succeed across different conditions, such as wet cartons, changing object weights, and varying surface friction.
